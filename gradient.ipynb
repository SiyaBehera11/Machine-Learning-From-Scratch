```python
# minimization of gradient descent for f(x) = x^2

def f(x):
    return x**2
def df(x):
    return 2*x
def gradient_descent(start_x, learning_rate, iterations):
    x = start_x
    
    for i in range(iterations):
        grad = df(x)
        x = x- learning_rate * grad

        print(f"Iteration {i}: x = {x}, f(x) = {f(x)}")
    return x
learning_rate = 0.1
start_x = 10
iterations = 10
final_x = gradient_descent(start_x, learning_rate, iterations)
print(f"Final x: {final_x}, f(x) = {f(final_x)}")
```

```python
#gradient decent using the slope formula 
def f(x):
    return x**2
def df(x):
    return 2*x
def numerical_decent(x, h=1e-5, learning_rate=0.1, iterations=12):
    for i in range(12):
        grad = (f(x + h) - f(x - h)) / (2 * h)  # Numerical approximation of the derivative
        x = x - learning_rate * grad
        print(f"Iteration {i}: x = {x}, f(x) = {f(x)}")
    return x
final_x = numerical_decent(start_x, learning_rate=learning_rate, iterations=iterations)
print(f"Final x: {final_x}, f(x) = {f(final_x)}")
```

```python
#gradient decent for 2 variables. f(x) = x^2 + y^2
def f(x,y):
    return x**2 + y**2
def df(x,y):
    return (2*x, 2*y)
def gradient_descent_2d(start_x, start_y, learning_rate, iterations):
    x, y = start_x, start_y
    for i in range(iterations):
        grad_x, grad_y = df(x,y)
        x = x- learning_rate * grad_x
        y = y -learning_rate * grad_y
        print(f"iterations {i}: x = {x}, y = {y}, f(x,y) = {f(x,y)}")
    return x, y
start_x, start_y = 3,4
iterations = 10
learning_rate = 0.1
final_x, final_y = gradient_descent_2d(start_x, start_y, learning_rate, iterations)
print(f"Final x: {final_x}, Final y: {final_y}, f(x,y) = {f(final_x, final_y)}")
```

```python
#linear regression using gradient descent using numpy
import numpy as np
X = np.array([1,2,3,4,5,6])
Y = np.array([25, 35, 45, 55, 65, 75])
class LinearRegression:
    def __init__(self, epochs, learning_rate):
        self.X = X
        self.Y = Y
        self.epochs = epochs
        self.learning_rate = learning_rate
        self.m = 0
        self.b = 0

    def predict(self, X):
        return self.m * X + self.b
    def calculate_loss(self):
        Y_pred = self.predict(self.X)
        return np.mean((self.Y - Y_pred) **2)
    def gradient_descent(self):
        n = len(self.X)
        for i in range(self.epochs):
            Y_pred = self.predict(self.X)
            d_m = (-2/n) * np.sum(self.X * (self.Y - Y_pred))
            d_b = (-2/n) * np.sum(self.Y - Y_pred)
            self.m = self.m - self.learning_rate * d_m
            self.b = self.b - self.learning_rate * d_b
            loss = self.calculate_loss()
            print(f"Epoch {i}: m = {self.m}, b = {self.b}, loss = {loss}")

    def fit(self, X, Y):
        self.X = X
        self.Y = Y
        self.gradient_descent()
model = LinearRegression(epochs=2000, learning_rate=0.01)
model.fit(X, Y)
print("Final weight:", model.m)
print("Final bias:", model.b)
print("Final loss:", model.calculate_loss())
```

```python
#House price prediction using gradient descent
import numpy as np
X = np.array ([500, 700, 900, 1100, 1300, 1500, 1700, 1900]) #house size in sq ft
Y = np.array([25, 35, 45, 55, 65, 75, 85, 95]) #house prices in lakhs
class HousePricePrediction:
    def __init__ (self, learning_rate, epochs):
        self.learning_rate = learning_rate
        self.epochs= epochs
        self.m = 0
        self.b = 0

    def predict(self, X):
        return self.m *X + self.b
    def calculate_loss(self):
        Y_pred = self.predict(self.X)
        return np.mean((self.Y - Y_pred) ** 2)
    def gradient_descent(self):
        n = len(self.X)
        for i in range(self.epochs):
            Y_pred = self.predict(self.X)
            d_m = (-2/n) * np.sum(self.X * (self.Y - Y_pred))
            d_b = (-2/n) * np.sum(self.Y - Y_pred)
            self.m -= self.learning_rate * d_m
            self.b -= self.learning_rate * d_b
            loss = self.calculate_loss()
            print(f"Epoch {i}: m = {self.m}, b = {self.b}, loss = {loss}") 

    def fit(self, X, Y):
        self.mean = np.mean(X)
        self.std = np.std(X)
        self.X = (X - self.mean) / self.std
        self.Y = Y
        self.gradient_descent()

model = HousePricePrediction(learning_rate=0.01, epochs=1000)
model.fit(X, Y)
print("Final weight:", model.m) 
print("Final bias:", model.b)
print("Final loss:", model.calculate_loss())
```

```python
# Non-linear regression using gradient descent
import numpy as np
X = np.array([-3, -2, -1, 0, 1, 2, 3])
Y = np.array([14, 7, 4, 5, 10, 19, 32])
class NonLinearRegression:
    def __init__ (self, learning_rate, epochs):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.a = 0
        self.b = 0
        self.c = 0

    def predict(self, X):
        return self.a * X**2 + self.b * X + self.c
    def calculate_loss (self):
        Y_pred = self.predict(self.X)
        return np.mean((self.Y - Y_pred) ** 2)
    def gradient_descent(self):
        n = len(self.X)
        for i in range(self.epochs):
            Y_pred = self.predict(self.X)
            d_a = (-2/n) * np.sum(self.X**2 * (self.Y - Y_pred))
            d_b = (-2/n) * np.sum(self.X * (self.Y - Y_pred))
            d_c = (-2/n) * np.sum(self.Y - Y_pred)
            self.a -= self.learning_rate * d_a
            self.b -= self.learning_rate * d_b
            self.c -= self.learning_rate * d_c
            loss = self.calculate_loss()
            print(f"Epoch {i}: a = {self.a}, b = {self.b}, c = {self.c}, loss = {loss}")
    def fit(self, X, Y):
        self.X = X
        self.Y = Y
        self.gradient_descent()
model = NonLinearRegression(learning_rate=0.01, epochs=500)
model.fit(X, Y)
print("Final a:", model.a)
print("Final b:", model.b)
print("Final c:", model.c)
print("Final loss:", model.calculate_loss())
X_test = np.array([-4, 4, 5, 6])
Y_pred = model.predict(X_test)
print("Predictions for X_test:", Y_pred)
```

```python
#logistic regression using gradient descent
import numpy as np
X = np.array([1,2,3,4,5,6,7,8])
Y = np.array([0,0,0,0,1,1,1,1])  # Binary labels
class LogisticRegression:
    def __init__(self, learning_rate, epochs):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.m = 0
        self.b = 0

    def sigmoid(self, z):
        return 1/ (1 + np.exp(-z))
    def predict(self, X):
        return self.sigmoid(self.m * X + self.b)
    def calculate_loss(self):
        Y_pred = self.predict(self.X)
        return -np.mean(self.Y * np.log(Y_pred) + (1 - self.Y) * np.log(1 - Y_pred))
    def gradient_descent(self):
        n = len(self.X)
        for i in range(self.epochs):
            Y_pred = self.predict(self.X)
            d_m = (-1/n) * np.sum(self.X * (self.Y - Y_pred))
            d_b = (-1/n) * np.sum(self.Y - Y_pred)
            self.m -= self.learning_rate * d_m
            self.b -= self.learning_rate * d_b
            loss = self.calculate_loss()
            print(f"Epoch {i}: m = {self.m}, b = {self.b}, loss = {loss}")

    def fit(self, X, Y):
        self.X = X
        self.Y = Y
        self.gradient_descent()
    def evaluate(self, X_test, Y_test):
        Y_pred = self.predict(X_test)
        Y_pred_labels = (Y_pred >= 0.5).astype(int)
        accuracy = np.mean(Y_pred_labels == Y_test)
        return accuracy
model = LogisticRegression(learning_rate=0.01, epochs=5000)
model.fit(X, Y)
model.evaluate(X, Y)
print("Final weight:", model.m) 
print("Final bias:", model.b)
print("Final loss:", model.calculate_loss())
X_test = np.array([2, 4, 5, 7])

probabilities = model.predict(X_test)

predictions = (probabilities >= 0.5).astype(int)

print("Probabilities:", probabilities)
print("Predicted classes:", predictions)
```
